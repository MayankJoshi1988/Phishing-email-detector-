# Phishing Email Detector
A beginner-friendly rule-based cybersecurity project. Run the next cell, then call `analyze_message(...)` with a message to inspect it.

```python
"""
Phishing Email Detector
A beginner-friendly, rule-based cybersecurity project.
Run: python phishing_email_detector.py
"""

import re

SUSPICIOUS_WORDS = [
    "urgent", "verify", "verification", "password", "account suspended",
    "click here", "login", "limited time", "bank details", "update your account",
    "confirm your account", "prize", "winner", "free gift", "immediately"
]

URL_PATTERN = re.compile(r"https?://\S+|www\.\S+", re.IGNORECASE)
IP_URL_PATTERN = re.compile(r"https?://(?:\d{1,3}\.){3}\d{1,3}", re.IGNORECASE)

def analyze_message(message):
    """Return a risk score, findings, and a simple risk label."""
    text = message.lower()
    findings = []
    score = 0

    matched_words = [word for word in SUSPICIOUS_WORDS if word in text]
    if matched_words:
        score += min(10, len(matched_words) * 2)
        findings.append("Suspicious wording: " + ", ".join(matched_words))

    urls = URL_PATTERN.findall(message)
    if urls:
        score += 2
        findings.append(f"Contains {len(urls)} link(s). Check links carefully before opening.")

    if IP_URL_PATTERN.search(message):
        score += 3
        findings.append("A link uses an IP address instead of a normal domain name.")

    if "@" in message and any(word in text for word in ["support", "security", "bank", "admin"]):
        score += 1
        findings.append("An email-like address or account-related wording was found; verify the sender.")

    score = min(score, 10)
    if score >= 7:
        label = "HIGH RISK"
    elif score >= 4:
        label = "MEDIUM RISK"
    else:
        label = "LOWER RISK"

    if not findings:
        findings.append("No obvious indicators were detected by these simple rules.")

    return {"score": score, "label": label, "findings": findings}

def main():
    print("=" * 48)
    print("        PHISHING EMAIL DETECTOR")
    print("=" * 48)
    print("Paste a suspicious email or message below.")
    print("Type END on a new line to analyze it.\n")

    lines = []
    while True:
        try:
            line = input()
        except EOFError:
            break
        if line.strip().upper() == "END":
            break
        lines.append(line)

    result = analyze_message("\n".join(lines))
    print("\n--- Analysis Result ---")
    print(f"Risk label: {result['label']}")
    print(f"Risk score: {result['score']}/10")
    print("Findings:")
    for item in result["findings"]:
        print(f"- {item}")
    print("\nSafety note: This is a basic educational tool, not a guarantee.")
    print("Do not click suspicious links or share passwords/OTP details.")

if __name__ == "__main__":
    main()
```

```python
sample = "URGENT! Your account is suspended. Verify your password now at http://192.168.1.5/login"
analyze_message(sample)
```
